# Visualizaciones

## Introduccion.

La etapa de visualización constituye un componente fundamental dentro del análisis exploratorio de datos, ya que permite transformar información numérica y categórica en representaciones gráficas que facilitan la interpretación de patrones, tendencias y relaciones presentes en el dataset. Después de haber depurado y filtrado la base de datos para conservar únicamente los registros correspondientes a diabetes mellitus tipo 2 (E11), y tras examinar su estructura, distribución temporal, territorial y demográfica, la visualización se convierte en el siguiente paso natural para profundizar en el comportamiento de esta causa de muerte en México.

Las gráficas permiten observar de manera intuitiva cómo evolucionan las defunciones a lo largo del tiempo, qué entidades federativas concentran mayor mortalidad, cómo se distribuyen los fallecimientos entre distintos grupos de edad y si existen diferencias relevantes entre hombres y mujeres. En esta sección se desarrollarán visualizaciones enfocadas en cuatro dimensiones clave del fenómeno estudiado:

 + Distribución territorial, mediante mapas y comparaciones entre entidades federativas.

 + Comportamiento por edad, analizando grupos etarios y su relación con la mortalidad.

Estas visualizaciones permitirán consolidar el análisis exploratorio y sentar las bases para la interpretación final del comportamiento de la diabetes tipo 2 como causa de muerte en México.

## Introducción a las tendencias temporales

El análisis de tendencias temporales es una parte esencial dentro del estudio de la mortalidad por diabetes mellitus tipo 2, ya que permite comprender cómo evoluciona esta causa de muerte a lo largo del tiempo y detectar patrones que no son evidentes en la inspección tabular del dataset. A través de la observación de variaciones anuales y mensuales, es posible identificar incrementos recientes, periodos de mayor incidencia, comportamientos estacionales y posibles cambios asociados a factores externos como políticas de salud, variaciones en la atención médica o modificaciones en los sistemas de registro.

En el caso del dataset filtrado, que contiene exclusivamente defunciones clasificadas bajo el código CIE‑10 E11, las variables anio_ocur y mes_ocurr proporcionan una base sólida para examinar la dinámica temporal de la mortalidad por diabetes tipo 2 en México. Estas tendencias permiten contextualizar la magnitud del problema, evaluar si la mortalidad ha aumentado o disminuido en los años más recientes y explorar si existen meses con mayor concentración de casos.

La visualización de estas tendencias temporales constituye un paso fundamental para transformar los datos en información interpretable, facilitando la identificación de patrones que servirán como base para análisis más profundos y para la discusión de los resultados en el contexto epidemiológico nacional.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Cargar datos
df_diabetes = pd.read_csv("../data/processed/defunciones_diabetes_tipo2.csv")

# Agrupar por año
df_year = df_diabetes.groupby("anio_ocur").size().reset_index(name="defunciones")

# Gráfica
plt.figure(figsize=(10,5))
plt.plot(df_year["anio_ocur"], df_year["defunciones"], marker="o", linewidth=2, color="#0077b6")
plt.title("Defunciones por Diabetes Tipo 2 por Año")
plt.xlabel("Año")
plt.ylabel("Número de defunciones")
plt.grid(True, alpha=0.3)
plt.show()
```
