# Mortalidad por diabetes en Mexico

## Introducción

La diabetes mellitus tipo 2 representa uno de los principales problemas de salud pública en México, tanto por su alta prevalencia como por su impacto en la mortalidad. En este contexto, el análisis sistemático de los registros de defunciones permite comprender la evolución del fenómeno, identificar patrones territoriales y temporales, y apoyar la toma de decisiones en materia de prevención y atención médica.

El Instituto Nacional de Estadística y Geografía (INEGI), en coordinación con la Secretaría de Salud (SSA), pone a disposición bases de datos que registran las causas de muerte en el país, clasificadas conforme a estándares internacionales como la CIE-10. Estos datos constituyen una fuente confiable y estructurada para realizar análisis exploratorios que permitan evaluar la magnitud de las muertes asociadas a la diabetes tipo 2 y su distribución por entidad federativa y año.

Este proyecto tiene como objetivo realizar un Análisis Exploratorio de Datos (EDA) sobre las defunciones por diabetes mellitus tipo 2 en México, utilizando información publicada por el INEGI. La exploración se centrará en:

 + Identificar tendencias temporales en la mortalidad.

 + Comparar la incidencia entre entidades federativas.

 + Detectar variaciones relevantes y posibles anomalías en los registros.

 + Generar visualizaciones que faciliten la interpretación de los datos.

El análisis busca ofrecer una visión clara y fundamentada del comportamiento de esta causa de muerte en el país, proporcionando insumos útiles para investigaciones posteriores, diseño de políticas públicas y estrategias de intervención en salud.

## Metodología del EDA

El análisis exploratorio se desarrollará siguiendo una secuencia lógica de etapas:

1. Obtención y carga del dataset
